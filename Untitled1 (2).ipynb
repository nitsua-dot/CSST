# **13. Create A Vizualization**

```r
students_risk <- function(attendance,grade){
  if (attendance <75 && grade <75){
    return("High risk")
  } else if (attendance <85 || grade<80){
    return("Moderate Risk")
  } else if (attendance >= 85 && grade >= 80){
    return("Low risk")
  }
  return("Unknown risk") # Default return for cases not explicitly covered
}

students$Risk <- mapply(
  students_risk,
  students$Attendance,
  students$Grade
)

risk_counts <- table(students$Risk)
barplot(
  risk_counts,
  main = "Student Risk Classification",
  xlab = "Risk Level",
  ylab = "Number of students"
)

print("1. There are more low and high risk students than moderate risk students")
print("1. Both are high and low risk")
print("1. Encourage high risk students to help the low risk students")
```
