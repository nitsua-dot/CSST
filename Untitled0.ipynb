Vectors

```r
grades <- c(98, 80,91,88,96)

grades
cat("Mean", mean(grades))
cat("Max", max(grades))
cat("Min", min(grades))
cat("Length", length(grades))
cat("Median",median(grades))
cat("Summary",summary(grades))
```
