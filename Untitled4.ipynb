# **Table 1. Debate Arguments**

```r
arguments <- data.frame(
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006", "A007", "A008"),
  Position = c("For", "For", "For", "Against", "Against", "Against", "Against", "Against"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assesment", "Dependency", "Dependency"),
  Claim = c(
    "AI can provide students with immediate learning support",
    "AI can help students complete routine academic tasks more efficiently",
    "Allowing AI use can help students develop AI literacy",
    "Undisclosed AI generated work may create academic intergrity concerns",
    "Excessive AI use may reduce students' independent problem solving practice",
    "AI use may make it more difficult to determine whether submitted work reflects student learning",
    "AI hinders the students critical and logical thinking",
    "Weaker memory and struggling to apply information"

  )
)
arguments
```

# **Table 2. Evidence**

```r
evidence <- data.frame(
  Evidence_ID = c("E001", "E002", "E003", "E004", "E005", "E006", "E007", "E008"),
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006", "E007", "E008"),
  Evidence_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research", "Research", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong", "Moderate", "Strong")

)
evidence
```

# **Connect Table Arguments and Evidence**

```r
knowledge_base <- merge(
  arguments,
  evidence,
  by = "Argument_ID"
)

knowledge_base
```

# **Convert Evidence Strenght into a Score**

```r
get_score <- function(strength){

  if (strength == "Strong"){
    return (3)
  } else if (strength == "Moderate"){
    return(2)
  } else if (strength == "Weak"){
    return(1)
  } else{
    return (0)
  }
}
```

# **Add evidence score**

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score

)

knowledge_base
```

# **Converting Evidence Score To Support**

```r
classify_support <- function(score){
  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else {
    return("No Support")
  }
}
classify_support(3)
```

# **Add Evidence Support**

```r
knowledge_base$Support_Level <- sapply(
  knowledge_base$Evidence_Score,
  classify_support
)
knowledge_base
```

# **Search By Topic**

```r
search_topic <- function(df,topic_name){
  result <- df[tolower(df$topic) == tolower(topic_name),]
  return(result)


}
```

```r
search_topic(knowledge_base, "Dependency")
```

# **Evidence Strength Visualization**

```r
visualize_strength <- function(df){
  barplot(
    table(df$strength),
    main = "evidence strength distribution",
    xlab = "strength",
    ylab = "count",
    col = c("skyblue", "orange", "lightgreen")
  )
}
```

```r
visualize_strength(knowledge_base)
```

# **Debate Report**

```r
generate_debate_report <- function(df){
  cat("final strength report \n \n")
  cat("total arguments", nrow(df), "\n")
  cat("for", sum(df$Position == "For"), "against", sum(df$Position == "Against"), "\n\n")
}
```

```r
generate_debate_report(knowledge_base)
```
